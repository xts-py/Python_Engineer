# Zadania: kolekcje i comprehensions

Kolejność:
1–2 listy, 3–4 słowniki, 5 krotki, 6 zbiory, 7–8 struktury złożone, 9 comprehensions, 10 zadanie łączone.

## Zadanie 1 – Playlista na imprezę
Poziom: rozgrzewka. Kolekcja: lista.

Masz początek playlisty w komórce poniżej.

1. Dopisz na koniec `"As It Was"`.
2. Wstaw na początek (indeks 0) `"Anti-Hero"`.
3. Usuń `"Shape of You"`.
4. Wypisz całą playlistę, liczbę utworów, pierwszy utwór, ostatni utwór i wycinek z dwoma pierwszymi.

Oczekiwany wynik:

```
['Anti-Hero', 'Blinding Lights', 'Levitating', 'As It Was']
Liczba utworów: 4
Pierwszy: Anti-Hero
Ostatni: As It Was
Początek: ['Anti-Hero', 'Blinding Lights']
```

Wskazówka: `append`, `insert`, `remove`, `len`, indeksy i wycinek `[:2]`.

## Zadanie 2 – Najlepsze rundy
Poziom: rozgrzewka. Kolekcja: lista.

W grze zebrałeś punkty z sześciu rund (lista w komórce).

Wypisz:
- sumę punktów,
- średnią zaokrągloną do jednego miejsca po przecinku,
- najlepszy i najgorszy wynik,
- ile razy wypadło dokładnie `150`,
- kopię listy posortowaną od największego do najmniejszego.

Oryginalna lista `rundy` ma zostać w starej kolejności. Wypisz ją na końcu, żeby to sprawdzić.

Oczekiwany wynik:

```
Suma: 650
Średnia: 108.3
Najlepszy: 150
Najgorszy: 60
Liczba 150: 2
Ranking: [150, 150, 120, 90, 80, 60]
Oryginał: [120, 80, 150, 90, 150, 60]
```

Wskazówka: `sum`, `len`, `round(..., 1)`, `max`, `min`, `count`, `sorted(..., reverse=True)`.

## Zadanie 3 – Karta postaci
Poziom: standard. Kolekcja: słownik.

1. Wypisz opis **przed walką**: `Shadow (mag) – poziom 12, HP: 80`
2. Boss zadaje 100 obrażeń. Odejmij je od HP. HP nie może spaść poniżej 0 (przy 80 HP i 100 obrażeniach zostaje 0, nie -20).
3. Awans: poziom rośnie o 1.
4. Dodaj klucz `"zloto"` z wartością `40`.
5. Wypisz tarczę przez `.get`. Gdy klucza nie ma, ma pojawić się słowo `brak`.
6. Wypisz cały słownik.

Oczekiwany wynik:

```
Shadow (mag) – poziom 12, HP: 80
Tarcza: brak
{'nick': 'Shadow', 'klasa': 'mag', 'poziom': 13, 'hp': 0, 'zloto': 40}
```

Wskazówka: opis z punktu 1 wypisz jeszcze przed zmianami. Do HP przyda się `max(0, ...)`.

## Zadanie 4 – Dziennik ocen
Poziom: standard. Kolekcja: słownik.

1. Dodaj `"angielski"` z oceną `5`.
2. Popraw wf z `3` na `4` (klucz już jest – wystarczy nadpisać wartość).
3. W pętli wypisz każdy przedmiot i ocenę w postaci `matematyka: 4`.
4. Oblicz średnią i wypisz ją.
5. Wypisz przedmioty z oceną co najmniej 5, oddzielone przecinkami.

Oczekiwany wynik:

```
matematyka: 4
polski: 5
informatyka: 6
wf: 4
angielski: 5
Średnia: 4.8
Co najmniej 5: polski, informatyka, angielski
```

Wskazówka: pętla `for przedmiot, ocena in oceny.items()`. Średnia to suma wartości podzielona przez liczbę przedmiotów.

## Zadanie 5 – Ruch na mapie
Poziom: standard. Kolekcja: krotka.

Pozycja gracza to krotka `(x, y)`.

1. Rozpakuj `pozycja` do zmiennych `x` i `y`. Wypisz `Jesteś na polu x=2, y=5`.
2. Gracz idzie w prawo o 3 i w górę o 1. Zbuduj **nową** krotkę `nowa_pozycja`. Starej nie zmieniaj. Wypisz obie.
3. Masz listę ruchów. Startujesz z `(0, 0)`.
   - `"prawo"` zwiększa x, `"lewo"` zmniejsza x,
   - `"gora"` zwiększa y, `"dol"` zmniejsza y.
   Każdy ruch to krotka `(kierunek, liczba_krokow)` – rozpakuj ją w pętli.
4. W komentarzu możesz sprawdzić, co zrobi `pozycja[0] = 10`. Program zgłosi błąd: krotki nie da się zmienić w miejscu. Zostaw tę linijkę zakomentowaną, żeby reszta komórki działała.

Oczekiwany wynik:

```
Jesteś na polu x=2, y=5
Nowa pozycja: (5, 6)
Stara pozycja: (2, 5)
Po ruchach: (1, 2)
```

Sprawdzenie ruchów z `(0, 0)`: prawo 2 → `(2, 0)`, góra 3 → `(2, 3)`, lewo 1 → `(1, 3)`, dół 1 → `(1, 2)`.

## Zadanie 6 – Wspólne gry
Poziom: standard. Kolekcja: zbiór.

Ola i Kuba zapisali, w co grają. Zbiór sam pilnuje, żeby gra nie powtórzyła się dwa razy.

Wypisz nazwy **posortowane alfabetycznie** i oddzielone przecinkami (`", ".join(sorted(...))`), żeby wynik zawsze wyglądał tak samo:

- gry, w które grają oboje,
- gry tylko Oli,
- gry tylko Kuby,
- wszystkie gry bez powtórzeń,
- czy `"Minecraft"` jest u Kuby (`tak` albo `nie`).

Oczekiwany wynik:

```
Wspólne: FIFA, Roblox
Tylko Ola: Among Us, Minecraft
Tylko Kuba: Fortnite, Valorant
Wszystkie: Among Us, FIFA, Fortnite, Minecraft, Roblox, Valorant
Minecraft u Kuby: nie
```

Wskazówka: `&` część wspólna, `-` różnica, `|` wszystko razem, `in` sprawdza, czy element jest w zbiorze.

## Zadanie 7 – Sklep z grami
Poziom: standard. Struktura: lista słowników.

Każda gra ma tytuł, cenę i gatunek.

1. Wypisz same tytuły, jeden pod drugim.
2. Policz, ile kosztują wszystkie gry razem.
3. Wypisz tytuły gier tańszych niż 100 zł (w jednej linii, po przecinku).
4. Znajdź najdroższą grę i wypisz jej tytuł oraz cenę.
5. Policz, ile gier jest z gatunku `"sandbox"`.

Oczekiwany wynik:

```
Minecraft
FIFA 25
Hades
Stardew Valley
Rocket League
Suma cen: 488
Taniej niż 100: Minecraft, Hades, Stardew Valley, Rocket League
Najdroższa: FIFA 25 (249 zł)
Sandbox: 2
```

Wskazówka: w pętli `gra` jest słownikiem, więc cena to `gra["cena"]`, a tytuł to `gra["tytul"]`. Najdroższą znajdziesz tak samo jak największą liczbę – tylko porównujesz `gra["cena"]`.

## Zadanie 8 – Ekwipunek drużyny
Poziom: standard. Struktura: słownik list.

1. Wypisz ekwipunek każdej osoby w postaci `Ola: miecz, tarcza, mikstura`.
2. **Dopiero potem** dodaj Mai przedmiot `"mikstura"`.
3. Wypisz, kto ma `"miecz"` (już po dodaniu mikstury – przy mieczu nic to nie zmienia).
4. Policz, ile mikstur jest w całej drużynie.
5. Zbuduj słownik: osoba → liczba przedmiotów. Wypisz go.

Oczekiwany wynik:

```
Ola: miecz, tarcza, mikstura
Kuba: luk, mikstura
Maja: miecz, zwoj
Kto ma miecz: Ola, Maja
Mikstur: 3
{'Ola': 3, 'Kuba': 2, 'Maja': 3}
```

Wskazówka: ekwipunek Mai to zwykła lista, więc `druzyna["Maja"].append(...)`. Mikstury policzysz przez `.count("mikstura")` na liście każdej osoby.

## Zadanie 9 – To samo, tylko krócej
Poziom: standard. Temat: comprehensions.

Zrób **wyłącznie comprehensions** (bez osobnej pętli `for`, która buduje wynik).
Każdy punkt wypisz osobnym `print`.

1. Lista ocen co najmniej 4. Zostaw powtórki – jeśli piątka jest dwa razy, ma być dwa razy.
2. Lista nicków wielkimi literami.
3. Słownik: nick → liczba liter.
4. Lista rund większych niż 100.
5. Zbiór gatunków ze sklepu z zadania 7. Kolejność w zbiorze może być inna. Mają być dokładnie: `akcja`, `sandbox`, `sport`.

Oczekiwany wynik (zbiór może wydrukować się w innej kolejności):

```
[5, 6, 4, 5]
['OLA', 'KUBA', 'MAJA', 'PIOTR']
{'ola': 3, 'kuba': 4, 'maja': 4, 'piotr': 5}
[120, 150]
{'akcja', 'sandbox', 'sport'}
```

Wskazówka: czytaj od `for`. Warunek dopisujesz na końcu: `[x for x in kolekcja if ...]`.

## Zadanie 10 – Turniej klasowy
Poziom: wyzwanie. Łączy listy, słowniki, zbiory i może użyć comprehension.

Każda osoba grała trzy mapy. Wyniki map są listą, a opis osoby słownikiem.

1. Dla każdej osoby policz sumę punktów i wypisz `Ola: 27` (w kolejności z listy `turniej`).
2. Wskaż zwycięzcę – osobę z największą sumą. W tych danych remisu nie ma.
3. Zbuduj słownik `nick → suma` i wypisz go. Może to być pętla albo dict comprehension.
4. Wypisz nicky osób z drużyny `"czerwoni"`, które mają sumę co najmniej 25.
5. Zbuduj zbiór nazw drużyn i wypisz je posortowane, oddzielone przecinkami.

Oczekiwany wynik:

```
Ola: 27
Kuba: 20
Maja: 29
Piotr: 17
Zwycięzca: Maja (29)
{'Ola': 27, 'Kuba': 20, 'Maja': 29, 'Piotr': 17}
Czerwoni z co najmniej 25: Ola, Maja
Drużyny: czerwoni, niebiescy
```

Wskazówka: suma jednej osoby to `sum(gracz["wyniki"])`. Zwycięzcę znajdziesz pętlą, tak jak największą liczbę – tylko pamiętasz też nick.
