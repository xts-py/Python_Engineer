## 2. Lista

Lista trzyma elementy **w kolejności**. Może mieć duplikaty. Da się ją zmieniać: dopisać, usunąć, podmienić element.

Indeksy działają jak w napisie: od `0`, a `-1` to ostatni element.
Wycinek `lista[start:stop]` **nie obejmuje** `stop` – ta sama zasada co w `range`.

````python
playlista = ["Blinding Lights", "Levitating", "As It Was", "Anti-Hero"]

print(playlista[0])     # pierwszy
print(playlista[-1])    # ostatni
print(playlista[1:3])   # od indeksu 1 do 2 (3 już nie wchodzi)
print("Levitating" in playlista)
````

````python
# Element listy można podmienić. W napisie się tak nie da.
playlista = ["Blinding Lights", "Levitating", "As It Was"]
playlista[0] = "Anti-Hero"
print(playlista)

napis = "Python"
# napis[0] = "J"   # TypeError – napis jest niemodyfikowalny
````

### Dopisywanie i usuwanie

- `append(x)` – dokłada **jeden** element na koniec
- `insert(indeks, x)` – wstawia w wybrane miejsce
- `extend(inna_lista)` – dokleja elementy drugiej listy (bez zagnieżdżania)
- `pop()` – zdejmuje ostatni element i go zwraca; `pop(i)` zdejmuje z indeksu `i`
- `remove(x)` – usuwa **pierwsze** wystąpienie wartości `x`

````python
playlista = ["Levitating", "As It Was"]
playlista.append("Anti-Hero")
playlista.insert(0, "Blinding Lights")
print(playlista)

zdjety = playlista.pop()
print("Zdjęty z końca:", zdjety)
print(playlista)

playlista.remove("Levitating")
print(playlista)
````

````python
# append kontra extend – częsta pomyłka
a = [1, 2]
a.append([3, 4])
print("append listy:", a)    # [1, 2, [3, 4]]  – cała lista jako jeden element

b = [1, 2]
b.extend([3, 4])
print("extend:", b)          # [1, 2, 3, 4]
````

### Sortowanie: `sort` zmienia listę, `sorted` robi nową

`lista.sort()` sortuje **w miejscu** i zwraca `None`.
`sorted(lista)` zostawia oryginał i zwraca nową listę.

````python
rundy = [120, 80, 150, 90, 60]

print("sort() zwraca:", rundy.sort())   # None – tu jest haczyk
print("po sort():", rundy)              # lista już jest posortowana

rundy = [120, 80, 150, 90, 60]
ranking = sorted(rundy, reverse=True)
print("oryginał:", rundy)
print("od największego:", ranking)
````

````python
# Przeglądanie: sama wartość albo numer + wartość
rundy = [120, 80, 150]
print("Suma:", sum(rundy))
print("Średnia:", round(sum(rundy) / len(rundy), 1))  # jedno miejsce po przecinku
print("Najlepszy:", max(rundy))
print("Najgorszy:", min(rundy))
print("Ile razy 150:", rundy.count(150))

playlista = ["Anti-Hero", "Levitating", "As It Was"]
for numer, utwor in enumerate(playlista, start=1):
    print(f"{numer}. {utwor}")
````

## 3. Słownik

Słownik łączy **klucz** z **wartością**. Szukamy po kluczu, nie po numerze miejsca.

Klucze się nie powtarzają. Wartości mogą. Od Pythona 3.7 słownik pamięta kolejność dodawania kluczy, ale i tak korzystamy z niego przez nazwę, nie przez indeks.

Dobre klucze to np. napisy, liczby i krotki. Lista kluczem być nie może.

````python
postac = {
    "nick": "Shadow",
    "klasa": "mag",
    "poziom": 12,
    "hp": 80,
}

print(postac["nick"])
print(f"{postac['nick']} ma {postac['hp']} HP")

postac["hp"] = 50          # zmiana istniejącego klucza
postac["zloto"] = 40       # nowego klucza po prostu nie było – teraz jest
print(postac)

# max działa też na dwóch liczbach, nie tylko na liście.
# Zostało 50 HP. Cios za 100 nie zbija HP poniżej zera.
obrazenia = 100
postac["hp"] = max(0, postac["hp"] - obrazenia)
print("HP po kolejnym ciosie:", postac["hp"])
````

````python
postac = {"nick": "Shadow", "hp": 80}

# print(postac["tarcza"])          # KeyError – klucza nie ma
print(postac.get("tarcza", "brak"))  # bezpiecznie: gdy nie ma klucza, wróć "brak"
print("hp" in postac)                # in sprawdza KLUCZE, nie wartości
print(80 in postac)                  # False
print(80 in postac.values())         # True
````

````python
oceny = {"matematyka": 4, "polski": 5, "informatyka": 6}

for przedmiot, ocena in oceny.items():
    print(f"{przedmiot}: {ocena}")

del oceny["polski"]
print(oceny)
````

````python
# zip spina dwie listy w pary. dict() zamienia pary na słownik.
# Gdy listy mają różną długość, zip kończy na krótszej.
nicki = ["Ola", "Kuba", "Maja"]
punkty = [27, 20, 29]
ranking = dict(zip(nicki, punkty))
print(ranking)
````

## 4. Krotka

Krotka jest jak lista, której **nie da się zmienić w miejscu**: nie dopiszesz elementu i nie podmienisz go przez indeks.

Używamy jej, gdy komplet jest stały: współrzędne `(x, y)`, kolor RGB `(255, 0, 0)`, para `(kierunek, liczba_krokow)`.

Rozpakowanie wkłada elementy krotki do osobnych zmiennych.

````python
pozycja = (2, 5)
x, y = pozycja
print(f"Jesteś na polu x={x}, y={y}")

# Ruch robi NOWĄ krotkę. Starej nie ruszamy.
nowa_pozycja = (x + 3, y + 1)
print("Nowa:", nowa_pozycja)
print("Stara:", pozycja)

# pozycja[0] = 9   # TypeError – krotka jest niemodyfikowalna
````

````python
# Krotka jednoelementowa POTRZEBUJE przecinka. Bez niego to zwykła wartość w nawiasie.
nie_krotka = (5)
krotka = (5,)
print(type(nie_krotka), type(krotka))
````

````python
# Dlatego krotka istnieje: może być kluczem słownika. Lista nie może.
skrzynie = {}
skrzynie[(2, 5)] = "złoto"
skrzynie[(0, 0)] = "mikstura"
print(skrzynie[(2, 5)])

# skrzynie[[2, 5]] = "złoto"   # TypeError: lista nie może być kluczem
````

> **Pytanie do grupy:** pozycja gracza na mapie – krotka czy lista?
>
> Jeśli traktujemy ją jako jedną stałą parę „tu stoisz” i ruch tworzy nową parę, bierzemy krotkę.
> Jeśli w trakcie gry ciągle podmieniamy `pozycja[0] = ...`, wygodniejsza jest lista. Oba wybory bywają dobre – ważne, żeby wiedzieć, co wolno.

## 5. Zbiór

Zbiór trzyma **unikalne** wartości. Nie ma indeksów i nie gwarantuje kolejności, więc nie piszemy `zbior[0]`.

Przydaje się do dwóch rzeczy: zdjęcia duplikatów oraz pytań „co wspólne / co tylko tu / co łącznie”.

Pusty zbiór to `set()`, nie `{}`. Puste `{}` to pusty **słownik**.

````python
ulubione = ["Minecraft", "FIFA", "Minecraft", "Roblox", "FIFA"]
unikalne = set(ulubione)
print(unikalne)          # kolejność może wyjść inna niż w liście
print(len(ulubione), "zapisów,", len(unikalne), "różnych gier")

print(type({}))          # dict
print(type(set()))       # set
````

````python
gry_oli = {"Minecraft", "FIFA", "Among Us", "Roblox"}
gry_kuby = {"FIFA", "Fortnite", "Roblox", "Valorant"}

print("Wspólne:   ", gry_oli & gry_kuby)     # część wspólna
print("Tylko Ola: ", gry_oli - gry_kuby)     # różnica
print("Tylko Kuba:", gry_kuby - gry_oli)
print("Wszystkie: ", gry_oli | gry_kuby)     # suma zbiorów

# To samo metodami: .intersection(), .difference(), .union()
print("Minecraft u Kuby:", "Minecraft" in gry_kuby)
````

````python
gry = {"Minecraft", "FIFA"}
gry.add("Hades")              # jeden element; odpowiednik append, ale słowo to add
gry.discard("nie ma takiej")  # nic się nie dzieje, gdy elementu nie ma
# gry.remove("nie ma takiej") # KeyError – remove wymaga, żeby element był
print(gry)
````

>
> Nie. Miejsca mają kolejność, a zbiór jej nie pilnuje. Ranking to lista (albo słownik miejsce → osoba).

### Dwie nazwy, jedna lista

`b = a` **nie kopiuje** listy. Obie zmienne wskazują ten sam obiekt.
Zmiana przez `b` widać też w `a`. Kopia to `a.copy()`.

````python
a = [1, 2, 3]
b = a
b.append(4)
print("a:", a)
print("b:", b)

c = a.copy()
c.append(5)
print("a po copy:", a)
print("c:", c)
````

`.copy()` kopiuje tylko **zewnętrzną** warstwę. Lista schowana w środku słownika nadal jest wspólna.
Do pełnej, niezależnej kopii służy `copy.deepcopy`.

Jeśli grupa jest już zmęczona, wystarczy poprzednia komórka (`b = a` kontra `.copy()`). Tę poniżej odpal, gdy ktoś zapyta, czemu kopia słownika nadal współdzieli listę w środku.

````python
import copy

oryginal = {"nick": "Ola", "itemy": ["miecz"]}
plytka = oryginal.copy()
plytka["nick"] = "Kuba"            # napis jest osobno – oryginał zostaje "Ola"
plytka["itemy"].append("tarcza")   # lista w środku jest WSPÓLNA

print("Płytka kopia:", plytka)
print("Oryginał:    ", oryginal)

oryginal = {"nick": "Ola", "itemy": ["miecz"]}
gleboka = copy.deepcopy(oryginal)
gleboka["itemy"].append("tarcza")
print("Po deepcopy, oryginał:", oryginal)
print("Głęboka kopia:       ", gleboka)
````

## 7. Comprehensions

Comprehension to krótki zapis pętli, która **buduje nową kolekcję**.

Czytaj od środka, od `for`:

```python
[ocena for ocena in oceny if ocena >= 4]
```

„Dla każdej oceny w ocenach, jeśli jest co najmniej 4, weź tę ocenę i zbierz z tego listę.”

````python
oceny = [5, 2, 6, 3, 4, 1, 5]

# Wersja, którą już znacie
dobre = []
for ocena in oceny:
    if ocena >= 4:
        dobre.append(ocena)
print(dobre)

# To samo w jednym wierszu
dobre = [ocena for ocena in oceny if ocena >= 4]
print(dobre)
````

````python
nicki = ["ola", "kuba", "maja", "piotr"]

wielkie = [nick.upper() for nick in nicki]
dlugosci = {nick: len(nick) for nick in nicki}

print(wielkie)
print(dlugosci)
````

````python
sklep = [
    {"tytul": "Minecraft", "cena": 99, "gatunek": "sandbox"},
    {"tytul": "FIFA 25", "cena": 249, "gatunek": "sport"},
    {"tytul": "Hades", "cena": 80, "gatunek": "akcja"},
    {"tytul": "Stardew Valley", "cena": 60, "gatunek": "sandbox"},
]

# Zbiór sam zdejmie powtórzone gatunki
gatunki = {gra["gatunek"] for gra in sklep}
print(gatunki)
````

### Dlaczego nie usuwać elementów w trakcie pętli

Pętla `for` idzie po kolei po pozycjach. Gdy skracasz listę w środku, kolejny element zostaje przeskoczony.
Zamiast usuwać – zbuduj nową listę. Comprehension robi dokładnie to.

````python
liczby = [1, 2, 4, 5]
for liczba in liczby:
    if liczba % 2 == 0:
        liczby.remove(liczba)
print("Usuwanie w pętli:", liczby)   # 4 zostaje – pętla ją przeskoczyła

liczby = [1, 2, 4, 5]
bez_parzystych = [liczba for liczba in liczby if liczba % 2 != 0]
print("Nowa lista:", bez_parzystych)
````

### Kiedy zwykła pętla jest lepsza

Comprehension ma zostać **krótkie i czytelne**.
Gdy w środku pojawia się kilka warunków, zagnieżdżenie albo efekt uboczny (`print`, dopisywanie do innej struktury) – napisz zwykłe `for`.

Dodatek (można pominąć na lekcji): comprehension w comprehension, czyli tabela z pętli w pętli.

````python
# Dodatek: tabliczka 1..3. Zewnętrzna pętla robi wiersze, wewnętrzna – kolumny.
tabliczka = [[a * b for b in range(1, 4)] for a in range(1, 4)]
for wiersz in tabliczka:
    print(wiersz)

# To samo bez comprehension, żeby było widać dwie pętle:
tabliczka = []
for a in range(1, 4):
    wiersz = []
    for b in range(1, 4):
        wiersz.append(a * b)
    tabliczka.append(wiersz)
print(tabliczka)
````

## 8. Pułapki – ściąga na koniec

Te sytuacje już się pojawiły. Warto je nazwać jeszcze raz:

1. **Indeks od 0**, wycinek i `range` nie obejmują końca.
2. **`sort()` zwraca `None`** i zmienia listę. Nowa lista: `sorted(...)`.
3. **`append(lista)` zagnieżdża.** Do doklejania elementów jest `extend`.
4. **`b = a` to nie kopia.** Do płaskiej listy wystarczy `.copy()`, przy zagnieżdżeniu `copy.deepcopy`.
5. **`slownik[klucz]` rzuca `KeyError`**, gdy klucza nie ma. Bezpiecznie: `.get(klucz, domyslna)`.
6. **`in` na słowniku sprawdza klucze**, nie wartości.
7. **`{}` to pusty słownik.** Pusty zbiór: `set()`.
8. **Krotka jednoelementowa** to `(x,)`. Bez przecinka nawias nic nie zmienia.
9. **Nie usuwaj z listy, po której właśnie idzie `for`.** Zbuduj nową listę.
10. Zbiór **nie ma kolejności** ani indeksu. Lista nie może być kluczem słownika ani elementem zbioru.
